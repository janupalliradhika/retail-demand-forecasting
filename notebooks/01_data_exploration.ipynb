## Day 4 - Data Cleaning and Transformation

```python
calendar = pd.read_csv("../data/calendar.csv")
sales = pd.read_csv("../data/sales_train_validation.csv")
prices = pd.read_csv("../data/sell_prices.csv")
```

```python
calendar["date"] = pd.to_datetime(calendar["date"])
```

```python
calendar_clean = calendar.copy()
sales_clean = sales.copy()
prices_clean = prices.copy()
```

```python
print("Calendar missing values:")
print(calendar_clean.isnull().sum())

print("\nPrices missing values:")
print(prices_clean.isnull().sum())
```

```python
calendar_clean["date"] = pd.to_datetime(calendar_clean["date"])
```

```python
calendar_clean["date"].dtype
```

```python
calendar_clean["year"] = calendar_clean["date"].dt.year
calendar_clean["month"] = calendar_clean["date"].dt.month
calendar_clean["week"] = calendar_clean["date"].dt.isocalendar().week
```

```python
calendar_clean[["date", "year", "month", "week"]].head()
```

## Day 7 - Store and Product Performance Analysis

```python
sales_clean.shape
```

```python
sales_columns = sales_clean.columns[6:]
```

```python
store_sales = sales_clean.groupby("store_id")[sales_columns].sum().sum(axis=1)
```

```python
store_sales.sort_values(ascending=False)
```

```python
plt.figure(figsize=(10,5))
store_sales.sort_values(ascending=False).plot(kind="bar")
plt.title("Total Sales by Store")
plt.xlabel("Store")
plt.ylabel("Total Units Sold")
plt.xticks(rotation=45)
plt.show()
```

```python
state_sales = sales_clean.groupby("state_id")[sales_columns].sum().sum(axis=1)
```

```python
state_sales.sort_values(ascending=False)
```

```python
plt.figure(figsize=(8,5))
state_sales.sort_values(ascending=False).plot(kind="bar")
plt.title("Total Sales by State")
plt.xlabel("State")
plt.ylabel("Total Units Sold")
plt.xticks(rotation=0)
plt.show()
```

```python
category_sales = sales_clean.groupby("cat_id")[sales_columns].sum().sum(axis=1)
```

```python
category_sales.sort_values(ascending=False)
```

```python
plt.figure(figsize=(8,5))
category_sales.sort_values(ascending=False).plot(kind="bar")
plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Units Sold")
plt.xticks(rotation=0)
plt.show()
```

```python
dept_sales = sales_clean.groupby("dept_id")[sales_columns].sum().sum(axis=1)
```

```python
dept_sales.sort_values(ascending=False).head(10)
```

```python
top_depts = dept_sales.sort_values(ascending=False).head(10)

plt.figure(figsize=(10,5))
top_depts.plot(kind="bar")
plt.title("Top 10 Departments by Sales")
plt.xlabel("Department")
plt.ylabel("Total Units Sold")
plt.xticks(rotation=45)
plt.show()
```

```python
item_sales = sales_clean.groupby("item_id")[sales_columns].sum().sum(axis=1)
```

```python
top_items = item_sales.sort_values(ascending=False).head(10)
top_items
```

```python
plt.figure(figsize=(12,5))
top_items.plot(kind="bar")
plt.title("Top 10 Products by Total Sales")
plt.xlabel("Product")
plt.ylabel("Total Units Sold")
plt.xticks(rotation=45)
plt.show()
```
